```python
#!/usr/bin/env python3
"""
Option Pain Theory Project
Author: Darshit Sarda

This script calculates the Option Pain (Max Pain) level for NIFTY/BANKNIFTY
based on option chain CSV files containing strike prices, call OI, and put OI.
"""

import os
import pandas as pd
import matplotlib.pyplot as plt


# ==========================
# CONFIGURATION
# ==========================
NIFTY_PATH = r"C:\Users\sarda\Desktop\monthly_options_view\index_options\NIFTY"
BANKNIFTY_PATH = r"C:\Users\sarda\Desktop\monthly_options_view\index_options\BANKNIFTY"


# ==========================
# FUNCTIONS
# ==========================

def read_option_chain(file_path: str) -> pd.DataFrame:
    """Read a CSV option chain file into a DataFrame."""
    df = pd.read_csv(file_path)

    # Adjust column names to match your CSV format
    df = df[['Strike', 'Call_OI', 'Put_OI']]
    df.rename(columns={'Strike': 'Strike Price', 
                       'Call_OI': 'Call OI', 
                       'Put_OI': 'Put OI'}, inplace=True)

    df = df.dropna()
    df = df.sort_values(by='Strike Price').reset_index(drop=True)

    return df


def calculate_total_loss(df: pd.DataFrame) -> pd.DataFrame:
    """
    For each strike, calculate the total loss to option writers (Call + Put).
    """
    strikes = df['Strike Price'].values
    call_oi = df['Call OI'].values
    put_oi = df['Put OI'].values

    total_losses = []

    for expiry_strike in strikes:
        call_loss = 0
        put_loss = 0

        # Call writers lose money if expiry > strike
        for strike, oi in zip(strikes, call_oi):
            if strike < expiry_strike:
                call_loss += (expiry_strike - strike) * oi

        # Put writers lose money if expiry < strike
        for strike, oi in zip(strikes, put_oi):
            if strike > expiry_strike:
                put_loss += (strike - expiry_strike) * oi

        total_loss = call_loss + put_loss
        total_losses.append(total_loss)

    df['Total Loss'] = total_losses
    return df


def find_max_pain(df: pd.DataFrame) -> int:
    """Return the strike price with minimum total loss."""
    min_loss_row = df.loc[df['Total Loss'].idxmin()]
    return int(min_loss_row['Strike Price'])


def plot_pain(df: pd.DataFrame, title: str, save_path: str = None):
    """Plot the Option Pain bar chart."""
    plt.figure(figsize=(10, 6))
    plt.bar(df['Strike Price'], df['Total Loss'], color='skyblue', edgecolor='black')
    plt.xlabel("Strike Price")
    plt.ylabel("Total Loss (to Option Writers)")
    plt.title(title)
    plt.xticks(rotation=45)

    # Highlight Max Pain
    max_pain = find_max_pain(df)
    plt.axvline(max_pain, color='red', linestyle='--', label=f"Max Pain = {max_pain}")
    plt.legend()

    if save_path:
        plt.savefig(save_path, bbox_inches="tight")
    plt.show()


def process_option_chain(file_path: str):
    """Read, process, and output results for a single option chain file."""
    df = read_option_chain(file_path)
    df = calculate_total_loss(df)
    max_pain = find_max_pain(df)

    print(f"\nFile: {os.path.basename(file_path)}")
    print(df[['Strike Price', 'Call OI', 'Put OI', 'Total Loss']])
    print(f"\n➡ Max Pain for {os.path.basename(file_path)}: {max_pain}")

    plot_pain(df, f"Option Pain - {os.path.basename(file_path)}")


def run_for_directory(directory: str, prefix: str):
    """Process all CSV files in a given directory starting with the given prefix."""
    for file in os.listdir(directory):
        if file.startswith(prefix) and file.endswith(".csv"):
            process_option_chain(os.path.join(directory, file))


# ==========================
# MAIN EXECUTION
# ==========================
if __name__ == "__main__":
    print("Running Option Pain Theory Project...\n")

    # Run for NIFTY
    print("===== NIFTY =====")
    run_for_directory(NIFTY_PATH, "NIFTY")

    # Run for BANKNIFTY
    print("\n===== BANKNIFTY =====")
    run_for_directory(BANKNIFTY_PATH, "BANKNIFTY")
```

```python
import pandas as pd
import numpy as np

def calculate_max_pain(file_path):
    # Read the CSV file
    df = pd.read_csv(file_path)
    
    # Sort by strike price to ensure proper calculation
    df = df.sort_values('Strike')
    
    # Extract necessary columns
    strikes = df['Strike'].values
    call_oi = df['Call_OI'].values
    put_oi = df['Put_OI'].values
    
    # Initialize list to store total loss for each candidate expiry
    total_losses = []
    
    # Calculate total loss for each strike considered as expiry
    for i, candidate in enumerate(strikes):
        loss_call = 0
        loss_put = 0
        
        # Calculate loss for call options (strike < expiry)
        for j in range(i + 1):
            if strikes[j] < candidate:
                loss_call += (candidate - strikes[j]) * call_oi[j]
        
        # Calculate loss for put options (strike > expiry)
        for k in range(i, len(strikes)):
            if strikes[k] > candidate:
                loss_put += (strikes[k] - candidate) * put_oi[k]
        
        total_losses.append(loss_call + loss_put)
    
    # Find the strike with minimum total loss
    min_loss_index = np.argmin(total_losses)
    max_pain_strike = strikes[min_loss_index]
    min_loss_value = total_losses[min_loss_index]
    
    # Create results DataFrame
    results_df = pd.DataFrame({
        'Strike': strikes,
        'Total Loss': total_losses
    })
    
    return max_pain_strike, min_loss_value, results_df

# File path to the CSV
file_path = r"C:\Users\sarda\Desktop\monthly_options_view\index_options\NIFTY\NIFTY_21-Aug-2025.csv"

# Calculate Max Pain
max_pain_strike, min_loss, loss_table = calculate_max_pain(file_path)

# Print results
print(f"Max Pain Strike: {max_pain_strike}")
print(f"Minimum Total Loss: {min_loss:,} points")
print("\nLoss Table Preview (First 5 Rows):")
print(loss_table.head())
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from typing import List, Dict, Optional

# ==========================
# CONFIGURATION
# ==========================
# Index input directories (use raw strings to avoid unicode escape issues)
NIFTY_PATH = r"C:\Users\sarda\Desktop\monthly_options_view\index_options\NIFTY"
BANKNIFTY_PATH = r"C:\Users\sarda\Desktop\monthly_options_view\index_options\BANKNIFTY"

# Equities root directory (contains many equity subfolders; exclude 'index_options')
EQUITIES_ROOT = r"C:\Users\sarda\Desktop\monthly_options_view"

# Toggle plotting
ENABLE_PLOTS = False

# ==========================
# UTILITIES
# ==========================

def _normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Normalize CSV column names to: Strike Price, Call OI, Put OI"""
    colmap = {c.lower().strip(): c for c in df.columns}
    candidates = {
        'strike': 'Strike Price',
        'strike price': 'Strike Price',
        'call_oi': 'Call OI',
        'call oi': 'Call OI',
        'ce oi': 'Call OI',
        'put_oi': 'Put OI',
        'put oi': 'Put OI',
        'pe oi': 'Put OI',
    }

    out = {}
    for key, target in candidates.items():
        if key in colmap:
            out[colmap[key]] = target

    required = {'Strike Price', 'Call OI', 'Put OI'}
    if set(out.values()) != required:
        missing = required - set(out.values())
        raise ValueError(
            f"CSV missing columns {missing}. Found: {list(df.columns)}"
        )

    df = df.rename(columns=out)
    for c in ['Strike Price', 'Call OI', 'Put OI']:
        df[c] = pd.to_numeric(df[c], errors='coerce')

    df = df.dropna(subset=['Strike Price', 'Call OI', 'Put OI'])
    df = df.sort_values(by='Strike Price').reset_index(drop=True)
    return df


def read_option_chain(file_path: str) -> pd.DataFrame:
    df = pd.read_csv(file_path)
    return _normalize_columns(df)


def calculate_total_loss(df: pd.DataFrame) -> pd.DataFrame:
    strikes = df['Strike Price'].values
    call_oi = df['Call OI'].values
    put_oi = df['Put OI'].values

    total_losses = []
    for expiry_strike in strikes:
        call_loss = ((expiry_strike - strikes).clip(min=0) * call_oi).sum()
        put_loss = ((strikes - expiry_strike).clip(min=0) * put_oi).sum()
        total_losses.append(call_loss + put_loss)

    df = df.copy()
    df['Total Loss'] = total_losses
    return df


def find_max_pain(df: pd.DataFrame) -> int:
    return int(df.loc[df['Total Loss'].idxmin()]['Strike Price'])


def plot_pain(df: pd.DataFrame, title: str, save_path: Optional[str] = None):
    plt.figure(figsize=(10, 6))
    plt.bar(df['Strike Price'], df['Total Loss'], edgecolor='black')
    plt.xlabel("Strike Price")
    plt.ylabel("Total Loss (to Option Writers)")
    plt.title(title)
    plt.xticks(rotation=45)

    max_pain = find_max_pain(df)
    plt.axvline(max_pain, linestyle='--', label=f"Max Pain = {max_pain}")
    plt.legend()

    if save_path:
        plt.savefig(save_path, bbox_inches="tight")
    if ENABLE_PLOTS:
        plt.show()
    else:
        plt.close()


def process_option_chain(file_path: str, do_plot: bool = False) -> Optional[int]:
    try:
        df = read_option_chain(file_path)
        if df.empty:
            print(f"⚠ Skipping empty file: {os.path.basename(file_path)}")
            return None

        df = calculate_total_loss(df)
        max_pain = find_max_pain(df)

        print(f"\nFile: {os.path.basename(file_path)}")
        print(df[['Strike Price', 'Call OI', 'Put OI', 'Total Loss']].to_string(index=False))
        print(f"➡ Max Pain: {max_pain}")

        if do_plot:
            title = f"Option Pain - {os.path.basename(file_path)}"
            plot_pain(df, title)

        return max_pain
    except Exception as e:
        print(f"❌ Failed to process {file_path}: {e}")
        return None


def run_for_directory(directory: str, prefix: str) -> None:
    if not os.path.isdir(directory):
        print(f"⚠ Directory not found: {directory}")
        return

    results = []
    for file in os.listdir(directory):
        if file.startswith(prefix) and file.lower().endswith(".csv"):
            file_path = os.path.join(directory, file)
            max_pain = process_option_chain(file_path, do_plot=ENABLE_PLOTS)
            if max_pain is not None:
                results.append({"File": file, "Max Pain": max_pain})

    if results:
        summary_df = pd.DataFrame(results).sort_values(by="File").reset_index(drop=True)
        summary_file = os.path.join(directory, f"{prefix}_MaxPain_Summary.csv")
        summary_df.to_csv(summary_file, index=False)
        print(f"\n📁 Saved summary to: {summary_file}")


def run_for_equity_folder(equity_dir: str) -> None:
    equity_name = os.path.basename(os.path.normpath(equity_dir))
    csv_files = [f for f in os.listdir(equity_dir) if f.lower().endswith(".csv")]

    if not csv_files:
        print(f"⚠ No CSV files in {equity_name}")
        return

    print(f"\n===== {equity_name} =====")
    results = []
    for file in sorted(csv_files):
        file_path = os.path.join(equity_dir, file)
        max_pain = process_option_chain(file_path, do_plot=ENABLE_PLOTS)
        if max_pain is not None:
            results.append({"File": file, "Max Pain": max_pain})

    if results:
        summary_df = pd.DataFrame(results)
        summary_path = os.path.join(equity_dir, f"{equity_name}_MaxPain_Summary.csv")
        summary_df.to_csv(summary_path, index=False)
        print(f"📁 Saved summary to: {summary_path}")


def run_for_all_equities(root_dir: str, exclude_dirs: Optional[List[str]] = None) -> None:
    if exclude_dirs is None:
        exclude_dirs = ['index_options']

    for name in sorted(os.listdir(root_dir)):
        full_path = os.path.join(root_dir, name)
        if not os.path.isdir(full_path):
            continue
        if name.lower() in {d.lower() for d in exclude_dirs}:
            continue
        run_for_equity_folder(full_path)


# ==========================
# MAIN
# ==========================
if __name__ == "__main__":
    print("Running Option Pain Theory Project...\n")

    print("===== NIFTY =====")
    run_for_directory(NIFTY_PATH, "NIFTY")

    print("\n===== BANKNIFTY =====")
    run_for_directory(BANKNIFTY_PATH, "BANKNIFTY")

    print("\n===== EQUITIES =====")
    run_for_all_equities(EQUITIES_ROOT, exclude_dirs=['index_options'])
```

```python
import os
import pandas as pd
from collections import defaultdict

# ==========================
# Config (same as your code)
# ==========================
EQUITIES_ROOT = r"C:\Users\sarda\Desktop\monthly_options_view"
EXCLUDE_DIRS = ['index_options']

OUTPUT_DIR = os.path.join(EQUITIES_ROOT, "Equity_Expiry_MaxPain")  # folder for expiry-wise summaries
os.makedirs(OUTPUT_DIR, exist_ok=True)


def collect_equity_max_pains(root_dir: str, exclude_dirs=None):
    """Scan all equity folders, collect max pain per expiry across equities."""
    if exclude_dirs is None:
        exclude_dirs = []

    expiry_dict = defaultdict(list)  # expiry -> list of dicts {Equity, MaxPain}

    for equity_name in sorted(os.listdir(root_dir)):
        equity_path = os.path.join(root_dir, equity_name)
        if not os.path.isdir(equity_path):
            continue
        if equity_name.lower() in {d.lower() for d in exclude_dirs}:
            continue

        # Read per-equity Max Pain summary file
        summary_file = os.path.join(equity_path, f"{equity_name}_MaxPain_Summary.csv")
        if not os.path.isfile(summary_file):
            print(f"⚠ Skipping {equity_name}, no summary file found")
            continue

        try:
            df = pd.read_csv(summary_file)
        except Exception as e:
            print(f"❌ Could not read {summary_file}: {e}")
            continue

        # Determine expiry from file names: assume File column has {equity}_{expiry}.csv
        for _, row in df.iterrows():
            file_col = row['File']
            max_pain = row['Max Pain']
            try:
                # extract expiry from file name
                if "_" in file_col:
                    parts = file_col.split("_")
                    expiry = parts[-1].replace(".csv", "")
                    expiry_dict[expiry].append({"Equity": equity_name, "Max Pain": max_pain})
            except Exception as e:
                print(f"⚠ Could not parse expiry from {file_col}: {e}")

    return expiry_dict


def write_expiry_wise_max_pain(expiry_dict, output_dir):
    """Write one CSV per expiry with columns Equity | Max Pain"""
    for expiry, records in expiry_dict.items():
        df = pd.DataFrame(records)
        df.sort_values("Equity", inplace=True)
        fname = f"Equity_MaxPain_{expiry}.csv".replace("/", "-")
        out_path = os.path.join(output_dir, fname)
        try:
            df.to_csv(out_path, index=False)
            print(f"✅ Wrote expiry summary: {out_path} ({len(df)} equities)")
        except Exception as e:
            print(f"❌ Could not write {out_path}: {e}")


if __name__ == "__main__":
    expiry_data = collect_equity_max_pains(EQUITIES_ROOT, exclude_dirs=EXCLUDE_DIRS)
    write_expiry_wise_max_pain(expiry_data, OUTPUT_DIR)
```
